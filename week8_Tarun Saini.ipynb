# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (TO IMPLEMENT)

def agent(query: str):
    query_lower = query.lower()

    try:
      if "calculate" in query_lower:
        expression=query_lower.replace("calculate","").strip()
        return{type:"calculation","result":calculator(expression)}
      elif "keyword" in query_lower:
        text=(query.replace("Extract keywords from","").replace("extract keywords from","").replace("keywords","").strip())
        return{"type":"keywords","result":extract_keywords(text)}
      else:
        return{"type":"general","result":f"I understand your question:'{query}'. This is a general query and does not require any tool."}
    except Exception as e:
      return{"type":"error","result":str(e)}
````

````python
import logging
import datetime

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

def word_count(text):
    return len(text.split())

def text_reverse(text):
    return text[::-1]

def get_current_time():
    return datetime.datetime.now().strftime("%d-%m-%Y %H:%M:%S")

def agent(query: str):
    query_lower = query.lower()
    logging.info(f"Processing user query: {query}")
    try:
        if any(word in query_lower for word in ["calculate", "add", "subtract", "multiply", "divide"]):
            expression = query_lower.replace("calculate", "").strip()
            logging.info("Routing to: Calculator")
            return {
                "tool": "Calculator",
                "status": "success",
                "result": calculator(expression)
            }

        elif "keyword" in query_lower or "extract" in query_lower:
            text = (query
                    .replace("Extract keywords from", "")
                    .replace("extract keywords from", "")
                    .replace("keywords", "")
                    .strip())
            logging.info("Routing to: Keyword Extractor")
            return {
                "tool": "Keyword Extractor",
                "status": "success",
                "result": extract_keywords(text)
            }

        elif "count words" in query_lower or "word count" in query_lower:
            text = query_lower.replace("count words", "").strip()
            logging.info("Routing to: Word Count Tool")
            return {
                "tool": "Word Counter",
                "status": "success",
                "result": word_count(text)
            }

        elif "reverse" in query_lower:
            text = query.replace("reverse", "").strip()
            logging.info("Routing to: Text Reverse Tool")
            return {
                "tool": "Text Reverse",
                "status": "success",
                "result": text_reverse(text)
            }

        elif "time" in query_lower or "current time" in query_lower:
            logging.info("Routing to: Current Time Tool")
            return {
                "tool": "Current Time",
                "status": "success",
                "result": get_current_time()
            }

        else:
            logging.info("Routing to: General Response")
            return {
                "tool": "General Response",
                "status": "success",
                "result": f"Query received: '{query}'. No specific tool matched, providing general acknowledgment."
            }

    except Exception as e:
        logging.error(f"Error processing query: {str(e)}")
        return {
            "tool": "Error Handler",
            "status": "failed",
            "result": str(e)
        }
````
